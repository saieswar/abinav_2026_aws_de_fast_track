```python
#parent/base/super Class
class A:
    
    def feature1(self):
        print("parent class Feature 1")
    
    def feature2(self):
        print("parent class Feature 2")

class B(A):
    
    def feature3(self):
        print("child feature 3")
    
    def feature4(self):
        print("child feature 4")
        

class C(B):
    def feature5(self):
        print("C class Feature 5")
```

```python
l = [1,5,11,4,5,1,1,4,2,2,2,9]
res = {}
count = 0
compare = l[0]
for i in l:
    if compare == i:
        count = count +1
res[compare]=count
res
```

```python
l = [1,5,11,4,5,1,1,4,2,2,2,9,1]
res = {}
for i in l:
    if i in  res:  #1 will not be in res
        res[i] = res[i] + 1
    else:
        res[i] = 1
# {1:4, 5:2, 11:1, 4: 1, 2:3,9:1 }
print(res)
```

```python
lst = [1, 2, 2, 3, 3, 3, 4, 4]

freq = {}

for i in lst:
    freq[i] = freq.get(i, 0) + 1

print(freq)

# {1:1,2:2 }

#freq[2] is nothing but freq.get(2, 0)
```
