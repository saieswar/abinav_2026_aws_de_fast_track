```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
from pyspark.sql.types import StructType, StructField, StringType
```

```python
spark = SparkSession.builder.appName("Filer Demo").master("local").getOrCreate()
```

```python
df = spark.read.format('csv').options(header=True).load("./inputData/StudentData.csv")   # extraction
```

```python
df.printSchema()
```

```python
df = df.filter( df.age > 28 )  #transformation 
```

```python
df.show()  #action
```

```python
df.where(df.name.contains('se')).show()
```

```python
df.filter(col("marks") > 50 ).show()
```

```python
df_marks_filtered = df.filter(col("marks") > 50 )
df_marks_filtered.count()
```

```python
df.createOrReplaceTempView("students")
df_sql= spark.sql("select *from students")
df_sql.show()
```

```python
df.filter((col("course") == "DB") & (col("marks")> 60) ).show()
```

```python

courses = ["Cloud", "OOP", "DSA"]
df.where( df.course.isin(courses) ).show()
```

```python
df.select("name", "age").show()
```

```python
df1 = spark.createDataFrame(data=data, schema=schema)
df1.show()
```
