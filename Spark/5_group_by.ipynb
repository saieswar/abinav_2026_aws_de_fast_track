```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg,max,count,sum,min 
```

```python
spark = SparkSession.builder.appName("group by demo").master("local").getOrCreate()
```

```python
df = spark.read.format("csv").options(header=True, inferSchema=True).load("./inputData/StudentData.csv")
df.printSchema()
```

```python
df.groupBy("gender").sum("marks").show()
```

```python
df.groupBy("gender").avg("marks").show()
```

```python
df1 = df.groupBy("gender").agg(
      sum("marks").alias("sum_salary"),\
      avg("marks").alias("avg_marks"),\
      max("marks").alias("max_marks"),\
      min("marks").alias("min_marks")
      )
df1.show()
```

```python
df2 = df.groupBy("gender", "age")\
        .agg(
            sum("marks").alias("sum_marks"),\
            avg("marks").alias("avg_marks"),\
            max("marks").alias("max_marks"),\
            min("marks").alias("min_marks")
        )
df2.show()
```

```python
df.createOrReplaceTempView("students")
spark.sql("select gender,age, sum(marks) as sum_marks from students group by gender, age").show()
```

```python
df.groupBy("course").agg(
    count("roll").alias("total_enroll")  #111 + 112
).show()
```

```python
df1.write.mode("append").format("csv").option("header", "true").save("./output/abhi_sch")
```
