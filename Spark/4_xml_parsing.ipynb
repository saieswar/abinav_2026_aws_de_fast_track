```python
from pyspark.sql import SparkSession

xml_data = """
<patient>
    <patientId>PAT1001</patientId>
    <name>
        <firstName>Ravi</firstName>
        <lastName>Kumar</lastName>
    </name>
    <age>35</age>
    <department>Cardiology</department>
</patient>
"""

spark = SparkSession.builder.getOrCreate()

data = [(xml_data,)]

df = spark.createDataFrame(data, ["xml_string"])

df.show(truncate=False)
```

```python
from pyspark.sql.functions import col, lit,xpath_string
df.select(
    xpath_string(col("xml_string"), lit("patient/patientId")).alias("patient_id")
).show(truncate=False)
```

```python
df.select(
    xpath_string(
        col("xml_string"),
        lit("patient/name/firstName")
    ).alias("first_name")
).show(truncate=False)
```

```python
result_df = df.select(
    xpath_string(col("xml_string"), lit("patient/patientId")).alias("patient_id"),
    xpath_string(col("xml_string"), lit("patient/name/firstName")).alias("first_name"),
    xpath_string(col("xml_string"), lit("patient/name/lastName")).alias("last_name"),
    xpath_string(col("xml_string"), lit("patient/age")).alias("age"),
    xpath_string(col("xml_string"), lit("patient/department")).alias("department")
)
result_df.show(truncate=False)
```

```python
df.select(
    xpath_string(col("xml_string"), lit("//age")).alias("age")
).show(truncate=False)
```

```python
xml_data = """
<patient patientType="OP">
    <patientId>PAT1001</patientId>
    <name>
        <firstName>Ravi</firstName>
        <lastName>Kumar</lastName>
    </name>
    <age>35</age>
    <department>Cardiology</department>
</patient>
"""
df = spark.createDataFrame(
    [(xml_data,)],
    ["xml_string"]
)
df.select(
    xpath_string(col("xml_string"), lit("patient/@patientType")).alias("patient_type")
).show(truncate=False)
```

namespace

```python
xml_data = """
<hc:patient xmlns:hc="http://example.com/healthcare">
    <hc:patientId>PAT1001</hc:patientId>
    <hc:name>
        <hc:firstName>Ravi</hc:firstName>
        <hc:lastName>Kumar</hc:lastName>
    </hc:name>
    <hc:age>35</hc:age>
    <hc:department>Cardiology</hc:department>
</hc:patient>
"""
df = spark.createDataFrame(
    [(xml_data,)],
    ["xml_string"]
)

from pyspark.sql.functions import xpath_string, col, lit

df.select(
    xpath_string(col("xml_string"), lit("patient/patientId")).alias("patient_id")
).show(truncate=False)
```

```python
df.select(
    xpath_string(col("xml_string"), lit("hc:patient/hc:patientId")).alias("patient_id")
).show(truncate=False)
```
