python dict to json string

```python
import json 
employee={
    "id": 101,
    "name": "Sai",
    "department": "Data Engineering",
    "skills": ["Python", "PySpark", "SQL"]
}
json_data = json.dumps(employee, indent=4) # converts python dict obj to json string
print(json_data)
type(json_data)
```

write json into a file

```python
import json 
employee = {
    "id": 101,
    "name": "Sai",
    "salary": 60000
}

with open("employees.json", "w") as file:
    json.dump(employee, file, indent=4)
```

```python
import json 
with open("employees.json", 'r') as file:
    employee = json.load(file)

print(type(employee))
print(employee)
print(employee["name"])
print(employee["salary"])
```

json str to dict

```python
import json 
json_string = '{"id": 101, "name": "Sai"}'
employee = json.loads(json_string)
print(employee)
print(type(employee))
```
