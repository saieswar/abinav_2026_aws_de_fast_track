```python
from lxml import etree

tree = etree.parse("ccd_001_valid.xml")
root = tree.getroot()

print(root)
print(root.tag)
```

```python
qname = etree.QName(root)  #?

print("Local name:", qname.localname)
print("Namespace:", qname.namespace)
```

```python
for element in root.iter():
    print(element.tag)
```

```python
for element in root.iter():
    if isinstance(element.tag, str):
        print(etree.QName(element).localname)
```

```python
ns = {"hc": "urn:hl7-org:v3"}

doc_id = root.xpath(
    "/hc:ClinicalDocument/hc:id/@extension",
    namespaces=ns
)

print(doc_id)
```

```python
member_id = root.xpath(
    "//hc:recordTarget/hc:patientRole/hc:id/@extension",
    namespaces=ns
)

print(member_id)
```

```python
first_name = root.xpath(
    "//hc:patient/hc:name/hc:given/text()",
    namespaces=ns
)

last_name = root.xpath(
    "//hc:patient/hc:name/hc:family/text()",
    namespaces=ns
)

dob = root.xpath(
    "//hc:patient/hc:birthTime/@value",
    namespaces=ns
)

gender = root.xpath(
    "//hc:patient/hc:administrativeGenderCode/@code",
    namespaces=ns
)

print(first_name)
print(last_name)
print(dob)
print(gender)
```

```python
titles = root.xpath(
    "//hc:section/hc:title/text()",
    namespaces=ns
)

print(titles)
```

```python
problems = root.xpath(
    "//hc:section[hc:title='Problems']//hc:observation/hc:value/@displayName",
    namespaces=ns
)

print(problems)
```

```python
lab_names = root.xpath(
    "//hc:section[hc:title='Results']//hc:observation/hc:code/@displayName",
    namespaces=ns
)

lab_values = root.xpath(
    "//hc:section[hc:title='Results']//hc:observation/hc:value/@value",
    namespaces=ns
)

lab_units = root.xpath(
    "//hc:section[hc:title='Results']//hc:observation/hc:value/@unit",
    namespaces=ns
)

print(lab_names)
print(lab_values)
print(lab_units)
```

```python
medication_names = root.xpath(
    "//hc:substanceAdministration//hc:manufacturedMaterial/hc:code/@displayName",
    namespaces=ns
)

dose = root.xpath(
    "//hc:substanceAdministration/hc:doseQuantity/@value",
    namespaces=ns
)

dose_unit = root.xpath(
    "//hc:substanceAdministration/hc:doseQuantity/@unit",
    namespaces=ns
)

route = root.xpath(
    "//hc:substanceAdministration/hc:routeCode/@displayName",
    namespaces=ns
)

print(medication_names)
print(dose)
print(dose_unit)
print(route)
```

```python
from lxml import etree
import glob

# Namespace
ns = {
    "hc": "urn:hl7-org:v3"
}

# Get all XML files from folder
xml_files = glob.glob("*.xml")

# Loop through each XML file
for file in xml_files:

    print("\nProcessing file:", file)

    # Parse XML
    tree = etree.parse(file)
    root = tree.getroot()

    # Extract Member ID
    member_id = root.xpath(
        "//hc:recordTarget/hc:patientRole/hc:id/@extension",
        namespaces=ns
    )

    # Extract First Name
    first_name = root.xpath(
        "//hc:patient/hc:name/hc:given/text()",
        namespaces=ns
    )

    # Extract Last Name
    last_name = root.xpath(
        "//hc:patient/hc:name/hc:family/text()",
        namespaces=ns
    )

    # Extract DOB
    dob = root.xpath(
        "//hc:patient/hc:birthTime/@value",
        namespaces=ns
    )

    # Extract Gender
    gender = root.xpath(
        "//hc:patient/hc:administrativeGenderCode/@code",
        namespaces=ns
    )

    # Print extracted information
    print("Member ID:", member_id)
    print("First Name:", first_name)
    print("Last Name:", last_name)
    print("DOB:", dob)
    print("Gender:", gender)
    # Convert XPath lists into individual values
    member_id = member_id[0] if member_id else None
    first_name = first_name[0] if first_name else None
    last_name = last_name[0] if last_name else None
    dob = dob[0] if dob else None
    gender = gender[0] if gender else None
```
